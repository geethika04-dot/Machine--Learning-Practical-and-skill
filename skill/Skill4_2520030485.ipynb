```python
# ============================================
# 1. IMPORT LIBRARIES
# ============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================
# 2. LOAD DATASET
# ============================================

df = pd.read_csv("Housing.csv")

print("Dataset loaded successfully!")
print()

print("First 5 rows:")
display(df.head())


# ============================================
# 3. BASIC INFORMATION
# ============================================

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()


# ============================================
# 4. CHECK MISSING VALUES
# ============================================

print("\nMissing Values:")
print(df.isnull().sum())


# ============================================
# 5. CHECK DUPLICATE VALUES
# ============================================

print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())


# ============================================
# 6. STATISTICAL SUMMARY
# ============================================

print("\nStatistical Summary:")
display(df.describe())


# ============================================
# 7. DATA PREPROCESSING
# ============================================

# Convert categorical variables into numerical values
df = pd.get_dummies(df, drop_first=True)

print("\nDataset after categorical encoding:")
display(df.head())


# ============================================
# 8. DEFINE FEATURES AND TARGET
# ============================================

X = df.drop("price", axis=1)

y = df["price"]

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget:")
print("price")


# ============================================
# 9. SPLIT DATA INTO TRAINING AND TESTING
# ============================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining data shape:")
print(X_train.shape)

print("\nTesting data shape:")
print(X_test.shape)


# ============================================
# 10. FEATURE SCALING
# ============================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("\nFeature scaling completed.")


# ============================================
# 11. LINEAR REGRESSION
# ============================================

linear_model = LinearRegression()

linear_model.fit(
    X_train_scaled,
    y_train
)

linear_pred = linear_model.predict(
    X_test_scaled
)


# ============================================
# 12. RIDGE REGRESSION
# L2 REGULARIZATION
# ============================================

ridge_model = Ridge(alpha=1.0)

ridge_model.fit(
    X_train_scaled,
    y_train
)

ridge_pred = ridge_model.predict(
    X_test_scaled
)


# ============================================
# 13. LASSO REGRESSION
# L1 REGULARIZATION
# ============================================

lasso_model = Lasso(
    alpha=0.01,
    max_iter=10000
)

lasso_model.fit(
    X_train_scaled,
    y_train
)

lasso_pred = lasso_model.predict(
    X_test_scaled
)


# ============================================
# 14. MODEL EVALUATION FUNCTION
# ============================================

def evaluate_model(model_name, actual, predicted):

    mae = mean_absolute_error(
        actual,
        predicted
    )

    mse = mean_squared_error(
        actual,
        predicted
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        actual,
        predicted
    )

    print("\n" + "=" * 45)
    print(model_name)
    print("=" * 45)

    print("MAE  :", round(mae, 2))
    print("MSE  :", round(mse, 2))
    print("RMSE :", round(rmse, 2))
    print("R²   :", round(r2, 4))

    return mae, mse, rmse, r2


# ============================================
# 15. EVALUATE LINEAR REGRESSION
# ============================================

linear_results = evaluate_model(
    "LINEAR REGRESSION",
    y_test,
    linear_pred
)


# ============================================
# 16. EVALUATE RIDGE REGRESSION
# ============================================

ridge_results = evaluate_model(
    "RIDGE REGRESSION",
    y_test,
    ridge_pred
)


# ============================================
# 17. EVALUATE LASSO REGRESSION
# ============================================

lasso_results = evaluate_model(
    "LASSO REGRESSION",
    y_test,
    lasso_pred
)


# ============================================
# 18. MODEL COMPARISON TABLE
# ============================================

results = pd.DataFrame(
    [
        linear_results,
        ridge_results,
        lasso_results
    ],
    columns=[
        "MAE",
        "MSE",
        "RMSE",
        "R2"
    ],
    index=[
        "Linear Regression",
        "Ridge Regression",
        "Lasso Regression"
    ]
)

print("\nMODEL COMPARISON")
display(results.round(4))


# ============================================
# 19. R² SCORE COMPARISON
# ============================================

plt.figure(figsize=(8, 5))

plt.bar(
    results.index,
    results["R2"]
)

plt.xlabel("Regression Model")
plt.ylabel("R² Score")
plt.title("R² Score Comparison")

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()


# ============================================
# 20. RMSE COMPARISON
# ============================================

plt.figure(figsize=(8, 5))

plt.bar(
    results.index,
    results["RMSE"]
)

plt.xlabel("Regression Model")
plt.ylabel("RMSE")
plt.title("RMSE Comparison")

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()


# ============================================
# 21. ACTUAL VS PREDICTED
# LINEAR REGRESSION
# ============================================

plt.figure(figsize=(8, 5))

plt.scatter(
    y_test,
    linear_pred,
    alpha=0.7
)

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")

plt.title(
    "Linear Regression - Actual vs Predicted"
)

plt.tight_layout()
plt.show()


# ============================================
# 22. ACTUAL VS PREDICTED
# RIDGE REGRESSION
# ============================================

plt.figure(figsize=(8, 5))

plt.scatter(
    y_test,
    ridge_pred,
    alpha=0.7
)

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")

plt.title(
    "Ridge Regression - Actual vs Predicted"
)

plt.tight_layout()
plt.show()


# ============================================
# 23. ACTUAL VS PREDICTED
# LASSO REGRESSION
# ============================================

plt.figure(figsize=(8, 5))

plt.scatter(
    y_test,
    lasso_pred,
    alpha=0.7
)

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")

plt.title(
    "Lasso Regression - Actual vs Predicted"
)

plt.tight_layout()
plt.show()


# ============================================
# 24. TEST DIFFERENT ALPHA VALUES
# ============================================

alphas = [
    0.01,
    0.1,
    1,
    10,
    100
]

ridge_scores = []
lasso_scores = []


for alpha in alphas:

    # Ridge
    ridge = Ridge(alpha=alpha)

    ridge.fit(
        X_train_scaled,
        y_train
    )

    ridge_prediction = ridge.predict(
        X_test_scaled
    )

    ridge_r2 = r2_score(
        y_test,
        ridge_prediction
    )

    ridge_scores.append(ridge_r2)


    # Lasso
    lasso = Lasso(
        alpha=alpha,
        max_iter=10000
    )

    lasso.fit(
        X_train_scaled,
        y_train
    )

    lasso_prediction = lasso.predict(
        X_test_scaled
    )

    lasso_r2 = r2_score(
        y_test,
        lasso_prediction
    )

    lasso_scores.append(lasso_r2)


# ============================================
# 25. ALPHA VS R² GRAPH
# ============================================

plt.figure(figsize=(8, 5))

plt.plot(
    alphas,
    ridge_scores,
    marker="o",
    label="Ridge"
)

plt.plot(
    alphas,
    lasso_scores,
    marker="o",
    label="Lasso"
)

plt.xscale("log")

plt.xlabel("Alpha")
plt.ylabel("R² Score")

plt.title(
    "Effect of Regularization Strength"
)

plt.legend()

plt.tight_layout()
plt.show()


# ============================================
# 26. FIND BEST MODEL
# ============================================

best_model = results["R2"].idxmax()

best_score = results["R2"].max()

print("\n" + "=" * 50)
print("BEST MODEL")
print("=" * 50)

print("Model :", best_model)
print("R² Score :", round(best_score, 4))


# ============================================
# 27. SAMPLE HOUSE PRICE PREDICTION
# ============================================

sample = X_test.iloc[[0]]

sample_scaled = scaler.transform(sample)

sample_prediction = ridge_model.predict(
    sample_scaled
)

print("\nSample House Details:")
display(sample)

print(
    "\nPredicted House Price using Ridge Regression:"
)

print(
    round(sample_prediction[0], 2)
)


# ============================================
# 28. FINAL CONCLUSION
# ============================================

print("\n" + "=" * 50)
print("PROJECT CONCLUSION")
print("=" * 50)

print("""
This project predicts house prices using Linear Regression,
Ridge Regression and Lasso Regression.

Linear Regression provides the basic prediction model.

Ridge Regression uses L2 regularization to reduce overfitting.

Lasso Regression uses L1 regularization and can perform
feature selection by reducing some coefficients toward zero.

The models are compared using MAE, MSE, RMSE and R² score.

The model with the highest R² score provides the best
performance among the tested models.
""")
```
